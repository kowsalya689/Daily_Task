```python
def create_one_hot_vector(word):
    one_hot_vector = np.zeros(vocabulary_size)
    one_hot_vector[word_to_index[word]] = 1
    return one_hot_vector
```

```python
for epoch_number in range(number_of_epochs):

    total_loss = 0

    for context_words, target_word in training_examples:

        context_vectors = get_context_vectors(context_words)

        average_context_vector = np.mean(
            context_vectors,
            axis=0
        )

        hidden_layer = np.dot(
            average_context_vector,
            input_weight_matrix
        )

        output_layer = np.dot(
            hidden_layer,
            output_weight_matrix
        )

        predicted_probabilities = softmax_function(
            output_layer
        )

        target_vector = create_one_hot_vector(
            target_word
        )

        loss = -np.sum(
            target_vector *
            np.log(predicted_probabilities + 1e-10)
        )

        total_loss += loss

        # Backpropagation
        prediction_error = (
            predicted_probabilities -
            target_vector
        )

        output_weight_gradient = np.outer(
            hidden_layer,
            prediction_error
        )

        hidden_layer_gradient = np.dot(
            output_weight_matrix,
            prediction_error
        )

        input_weight_gradient = np.outer(
            average_context_vector,
            hidden_layer_gradient
        )

        # Update weights
        output_weight_matrix -= (
            learning_rate *
            output_weight_gradient
        )

        input_weight_matrix -= (
            learning_rate *
            input_weight_gradient
        )

    print(
        "Epoch:",
        epoch_number + 1,
        "Loss:",
        round(total_loss, 4)
    )
```

```python
for epoch_number in range(number_of_epochs):

    total_loss = 0

    for context_words, target_word in training_examples:

        # Get one-hot vectors for context words
        context_vectors = get_context_vectors(context_words)

        # Average context vectors
        average_context_vector = np.mean(
            context_vectors,
            axis=0
        )

        # Forward pass
        hidden_layer = np.dot(
            average_context_vector,
            input_weight_matrix
        )

        output_layer = np.dot(
            hidden_layer,
            output_weight_matrix
        )

        # Softmax
        predicted_probabilities = softmax_function(
            output_layer
        )

        # Target vector
        target_vector = create_one_hot_vector(
            target_word
        )

        # Loss
        loss = -np.sum(
            target_vector *
            np.log(predicted_probabilities + 1e-10)
        )

        total_loss += loss

        # -------------------------
        # Backpropagation
        # -------------------------

        prediction_error = (
            predicted_probabilities -
            target_vector
        )

        output_weight_gradient = np.outer(
            hidden_layer,
            prediction_error
        )

        hidden_layer_gradient = np.dot(
            output_weight_matrix,
            prediction_error
        )

        input_weight_gradient = np.outer(
            average_context_vector,
            hidden_layer_gradient
        )

        # -------------------------
        # Update weights
        # -------------------------

        output_weight_matrix -= (
            learning_rate *
            output_weight_gradient
        )

        input_weight_matrix -= (
            learning_rate *
            input_weight_gradient
        )

    print(
        "Epoch:",
        epoch_number + 1,
        "Loss:",
        round(total_loss, 4)
    )
```

```python
import numpy as np

def cosine_similarity(vector1, vector2):

    numerator = np.dot(vector1, vector2)

    denominator = (
        np.linalg.norm(vector1) *
        np.linalg.norm(vector2)
    )

    if denominator == 0:
        return 0

    return numerator / denominator
```

```python
def find_similar_words(search_word, top_n=10):

    if search_word not in word_to_index:
        print(
            f"'{search_word}' is not available "
            "in the vocabulary."
        )
        return

    search_index = word_to_index[search_word]

    search_vector = input_weight_matrix[search_index]

    similarities = []

    for word, index in word_to_index.items():

        if word == search_word:
            continue

        word_vector = input_weight_matrix[index]

        similarity = cosine_similarity(
            search_vector,
            word_vector
        )

        similarities.append(
            (word, similarity)
        )

    similarities.sort(
        key=lambda x: x[1],
        reverse=True
    )

    return similarities[:top_n]
```

```python
# ============================================================
# SHERLOCK HOLMES WORD2VEC USING GENSIM - CBOW
# Complete Google Colab Code
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q -U gensim pypdf


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

from pypdf import PdfReader
from gensim.models import Word2Vec
import re
import os


# ============================================================
# 3. BUSINESS UNDERSTANDING
# ============================================================

project_name = "Sherlock Holmes Word2Vec Using Gensim"

business_objective = (
    "Learn semantic relationships between words "
    "using CBOW Word2Vec."
)

success_criteria = (
    "Given a word, return the most similar words."
)

print("=" * 60)
print(project_name)
print("=" * 60)

print("Business Objective:")
print(business_objective)

print("\nSuccess Criteria:")
print(success_criteria)


# ============================================================
# 4. PDF FILE PATH
# ============================================================

# Change this path if your PDF is stored somewhere else.

pdf_file_path = "/content/drive/MyDrive/Word Embeddings & Semantic Search/cano.pdf"

print("\nPDF Path:")
print(pdf_file_path)


# ============================================================
# 5. CHECK PDF EXISTS
# ============================================================

if not os.path.exists(pdf_file_path):

    raise FileNotFoundError(
        f"PDF file not found:\n{pdf_file_path}"
    )

print("\nPDF found successfully!")


# ============================================================
# 6. EXTRACT TEXT FROM PDF
# ============================================================

pdf_reader = PdfReader(pdf_file_path)

book_text = ""

for pdf_page in pdf_reader.pages:

    extracted_page_text = pdf_page.extract_text()

    if extracted_page_text:

        book_text += (
            extracted_page_text + "\n"
        )

print("\nPDF Information")
print("-" * 60)

print(
    "Total Pages:",
    len(pdf_reader.pages)
)

print(
    "Total Characters:",
    len(book_text)
)

print("\nFirst 500 characters:")
print(book_text[:500])


# ============================================================
# 7. TEXT CLEANING
# ============================================================

book_text = book_text.lower()

# Remove unwanted characters
cleaned_text = re.sub(
    r"[^a-zA-Z\s.!?]",
    " ",
    book_text
)

# Remove extra spaces
cleaned_text = re.sub(
    r"\s+",
    " ",
    cleaned_text
).strip()

print("\nCleaned Text")
print("-" * 60)
print(cleaned_text[:500])


# ============================================================
# 8. CREATE SENTENCES
# ============================================================

sentences = re.split(
    r"[.!?]+",
    cleaned_text
)

print(
    "\nTotal Raw Sentences:",
    len(sentences)
)


# ============================================================
# 9. TOKENIZE SENTENCES
# ============================================================

tokenized_sentences = []

for sentence in sentences:

    words = re.findall(
        r"\b[a-zA-Z]+\b",
        sentence
    )

    if len(words) > 1:

        tokenized_sentences.append(words)


print(
    "Total Tokenized Sentences:",
    len(tokenized_sentences)
)


# ============================================================
# 10. DISPLAY SAMPLE SENTENCES
# ============================================================

print("\nFirst 5 Tokenized Sentences")
print("-" * 60)

for sentence in tokenized_sentences[:5]:

    print(sentence)


# ============================================================
# 11. CHECK TOTAL TOKENS
# ============================================================

total_tokens = sum(
    len(sentence)
    for sentence in tokenized_sentences
)

print("\nTotal Tokens:", total_tokens)


# ============================================================
# 12. CBOW MODEL PARAMETERS
# ============================================================

embedding_size = 50

context_window_size = 2

min_word_frequency = 1

number_of_workers = 4

number_of_epochs = 50


print("\nCBOW Model Parameters")
print("-" * 60)

print(
    "Embedding Size:",
    embedding_size
)

print(
    "Context Window:",
    context_window_size
)

print(
    "Minimum Word Frequency:",
    min_word_frequency
)

print(
    "Workers:",
    number_of_workers
)

print(
    "Epochs:",
    number_of_epochs
)


# ============================================================
# 13. CREATE CBOW WORD2VEC MODEL
# ============================================================

cbow_model = Word2Vec(

    vector_size=embedding_size,

    window=context_window_size,

    min_count=min_word_frequency,

    workers=number_of_workers,

    sg=0,

    epochs=number_of_epochs
)


print("\nCBOW model architecture created!")


# ============================================================
# 14. BUILD VOCABULARY
# ============================================================

cbow_model.build_vocab(
    tokenized_sentences
)

print(
    "Vocabulary Size:",
    len(cbow_model.wv)
)

print(
    "Total Corpus Words:",
    cbow_model.corpus_total_words
)


# ============================================================
# 15. TRAIN CBOW MODEL
# ============================================================

print("\nStarting CBOW training...")

cbow_model.train(

    tokenized_sentences,

    total_examples=cbow_model.corpus_count,

    epochs=number_of_epochs

)

print("\nCBOW training completed successfully!")


# ============================================================
# 16. CHECK VOCABULARY
# ============================================================

print("\nFinal Vocabulary Size:")
print(len(cbow_model.wv))


# ============================================================
# 17. CHECK ONE WORD EMBEDDING
# ============================================================

test_word = "sherlock"

print("\nTesting Word Embedding")
print("-" * 60)

if test_word in cbow_model.wv:

    word_vector = cbow_model.wv[test_word]

    print(
        "Word:",
        test_word
    )

    print(
        "Embedding Size:",
        len(word_vector)
    )

    print(
        "Embedding Vector:"
    )

    print(word_vector)

else:

    print(
        f"'{test_word}' is not in the vocabulary."
    )


# ============================================================
# 18. FIND SIMILAR WORDS
# ============================================================

search_word = "sherlock"

print("\nSemantic Similarity Test")
print("-" * 60)

if search_word in cbow_model.wv:

    similar_words = cbow_model.wv.most_similar(

        search_word,

        topn=10

    )

    print(
        f"Top Similar Words for '{search_word}':\n"
    )

    for rank, (word, score) in enumerate(
        similar_words,
        start=1
    ):

        print(
            f"{rank}. "
            f"{word:<20} "
            f"Similarity: {score:.4f}"
        )

else:

    print(
        f"'{search_word}' is not in the vocabulary."
    )


# ============================================================
# 19. TEST ANOTHER WORD
# ============================================================

search_word = "watson"

print("\nAnother Semantic Similarity Test")
print("-" * 60)

if search_word in cbow_model.wv:

    similar_words = cbow_model.wv.most_similar(

        search_word,

        topn=10

    )

    print(
        f"Top Similar Words for '{search_word}':\n"
    )

    for rank, (word, score) in enumerate(
        similar_words,
        start=1
    ):

        print(
            f"{rank}. "
            f"{word:<20} "
            f"Similarity: {score:.4f}"
        )

else:

    print(
        f"'{search_word}' is not in the vocabulary."
    )


# ============================================================
# 20. SAVE THE TRAINED MODEL
# ============================================================

model_file = (
    "/content/sherlock_holmes_cbow_word2vec.model"
)

cbow_model.save(
    model_file
)

print("\nModel Saved Successfully!")
print(model_file)


# ============================================================
# 21. LOAD THE SAVED MODEL
# ============================================================

loaded_model = Word2Vec.load(
    model_file
)

print("\nSaved Model Loaded Successfully!")


# ============================================================
# 22. TEST THE LOADED MODEL
# ============================================================

search_word = "sherlock"

print("\nTesting Loaded Model")
print("-" * 60)

if search_word in loaded_model.wv:

    results = loaded_model.wv.most_similar(

        search_word,

        topn=10

    )

    print(
        f"Similar words for '{search_word}':\n"
    )

    for rank, (word, score) in enumerate(
        results,
        start=1
    ):

        print(
            f"{rank}. "
            f"{word:<20} "
            f"{score:.4f}"
        )

else:

    print(
        f"'{search_word}' not found."
    )


# ============================================================
# 23. INTERACTIVE SEMANTIC SEARCH
# ============================================================

print("\n" + "=" * 60)
print("INTERACTIVE SEMANTIC SEARCH")
print("=" * 60)

while True:

    search_word = input(
        "\nEnter a word "
        "(type 'exit' to stop): "
    ).lower().strip()

    if search_word == "exit":

        print(
            "\nSemantic search stopped."
        )

        break

    if search_word in loaded_model.wv:

        results = loaded_model.wv.most_similar(

            search_word,

            topn=10

        )

        print(
            f"\nTop Similar Words for "
            f"'{search_word}':"
        )

        print("-" * 50)

        for rank, (word, score) in enumerate(
            results,
            start=1
        ):

            print(
                f"{rank}. "
                f"{word:<20} "
                f"Similarity: {score:.4f}"
            )

    else:

        print(
            f"\n'{search_word}' is not "
            "available in the vocabulary."
        )


# ============================================================
# END OF PROJECT
# ============================================================

print("\n" + "=" * 60)
print("SHERLOCK HOLMES CBOW WORD2VEC PROJECT COMPLETED")
print("=" * 60)
```
